## **Samples Images**

Below, we display **5 sample images per class**, with each row corresponding to a different cancer type.

```python
import matplotlib.pyplot as plt

from PIL import Image
from supervised.src.utils import load_sample_filepath

def draw_sample_images(filepath: str):
    names, filepaths = load_sample_filepath(filepath)

    plt.figure(figsize=(10,5))

    for idx, name in enumerate(names):
        for j, filename in enumerate(filepaths[idx]):
            path = str(SUPERVISED_ROOT / "dataset" / "Training" / name / filename)
            img = Image.open(path)
            plt.subplot(len(names), 5, idx * 5 + j + 1)
            plt.imshow(img)
            plt.axis("off")
            if j == 2:
                plt.title(name)
    plt.suptitle("Sample images from each class")
    plt.tight_layout()
    plt.show()

draw_sample_images(CSV_PATH)
```

### **Results Interpretation**

The images appear consistent in terms of format and visual quality.<br>
Each class displays distinct visual patterns.

## **Dataset Structure**

The dataset is divided into a **training set** and a **testing set**.

```python
from supervised.src.utils import get_counts_split

def plot_distribution(filepath: str):
    count_training, count_testing = get_counts_split(filepath)

    labels = ["Training", "Testing"]
    values = [count_training, count_testing]

    plt.figure(figsize=(6,6))

    plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90, colors=['#1f77b4', '#ff7f0e'])

    plt.title("Dataset Split Distribution")

    plt.show()

if __name__ == "__main__":
    plot_distribution(CSV_PATH)
```

### **Results Interpretation**

The dataset is composed of **81.3% training data** and **18.7% testing data**.<br>
